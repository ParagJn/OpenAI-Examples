### Important - Setting up of .env variables to hide the keys from getting exposed and misused.
While it’s possible to hardcode your API key directly into your Python scripts, this is not recommended. Best practices suggest keeping sensitive information, such as API keys, separate from your codebase to enhance security. A common and effective method is to store the API key in a .env file and load it using the python-dotenv package. Here’s how to set it up:
1.	Create a .env File:
	- In the same directory as your notebook or script, create a file named .env.
2.	Add Your API Key:
	- Open the .env file and add your API key in the following format:
    
	**ANTHROPIC_API_KEY=your-api-key-here**
    
	Save the file after making this change.

3.	Install python-dotenv:
	- Run the following command in your terminal or notebook to install the python-dotenv package:

```python
#To install the package from a notebook: 
%pip install python-dotenv

#To install the package from your terminal:
# pip install python-dotenv
```

#### Lets Start - First connection with the model. Say Hello World

```python
# Connecting to Claude Model. 

from dotenv import load_dotenv
import os
from anthropic import Anthropic

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")

# Connect using the api
client = Anthropic(api_key=anthropic_api_key)

hello_world_message = client.message.create(
    model="claude-3-haiku-20240307",
    max_tokens=100,
    messages=[
        {"role": "user", "content": "Hello Claude - How can you help?"}
    ]
)
```
